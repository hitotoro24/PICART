```python
import numpy as np
from sklearn.tree import DecisionTreeRegressor

x = np.linspace(-1, 1, 70)
X = np.stack(np.meshgrid(x, x), -1).reshape(-1, 2)
Qtrue = np.array([[2., 1.], [1., 4.]])
y = .4*X[:,0] - .3*X[:,1] + .5*np.einsum('ni,ij,nj->n', X, Qtrue, X)
tree = DecisionTreeRegressor(max_depth=7, random_state=0).fit(X, y)
Qhat, Qraw, rows = estimate_tree_hessian(tree, X)
print('True Hessian:\n', Qtrue)
print('Raw tree rows:\n', np.round(Qraw, 3))
print('Algorithm 1 result:\n', np.round(Qhat, 3) if Qhat is not None else 'not identified')
print('Rows (axis, nodes, pairs, rank):', rows)
print('Relative error:', np.linalg.norm(Qhat-Qtrue)/np.linalg.norm(Qtrue) if Qhat is not None else None)
```
