## §2 Background on Physics-Informed Machine Learning (PIML)

<a id="eq-1"></a>
### Eq. (1) — Differential equation model
*§2, PDF p. 4*

$$
\mathcal{D}f(\mathbf{x}) - h(f(\mathbf{x}), \mathbf{x}) = 0 \;\; \forall \mathbf{x} \in \mathcal{X} \,, \tag{1}
$$

[↑ index](#eq-index)

<a id="eq-2"></a>
### Eq. (2) — Boundary / initial conditions
*§2, PDF p. 4*

$$
f(\mathbf{x}) = b(\mathbf{x}) \;\; \forall \mathbf{x} \in \mathcal{B} \subseteq \partial\mathcal{X} \,. \tag{2}
$$

[↑ index](#eq-index)

<a id="eq-3"></a>
### Eq. (3) — PIML penalized objective ★
*§2, PDF p. 4*

$$
\hat{f} = \underset{f \in \mathcal{F}}{\operatorname{argmin}} \sum_{i=1}^{N} (f(\mathbf{x}_i) - y_i)^2 + \lambda_d \int_{\mathcal{X}} (\mathcal{D}f(\mathbf{x}) - h(f(\mathbf{x}), \mathbf{x}))^2 d\mathbf{x} \,, + \lambda_b \int_{\mathcal{B}} (f(\mathbf{x}) - b(\mathbf{x}))^2 d\mathbf{x} \,, \tag{3}
$$

> ⚠️ **Transcription flags** (original kept as printed):
> - A comma appears before $+\lambda_b$ in the PDF ("$d\mathbf{x}\,, +\lambda_b$"); kept as printed. Eq. (7) has no such comma.

## §3.2 Computing the TBDA for Gradients

<a id="eq-4"></a>
### Eq. (4) — TBDA for gradients (node $i$) ★
*§3.2, PDF p. 6*

$$
\left.\frac{\tilde{\partial} f}{\tilde{\partial} x_p}\right|_i := \frac{2(\mu_i^r - \mu_i^l)}{u^i_{\sigma_i} - l^i_{\sigma_i}} \approx \frac{\partial f}{\partial x_{\sigma_i}}(\mathbf{x}) \;\; \forall \mathbf{x} \in [\mathbf{l}^i, \mathbf{u}^i] \,. \tag{4}
$$

> ⚠️ **Transcription flags** (original kept as printed):
> - Left side is indexed by $x_p$; right side by $x_{\sigma_i}$ (the text sets $\sigma_i = p$).
> - Algorithm 1 (same page) writes the denominator as $u_{i,\sigma_i} - l_{i,\sigma_i}$, whereas Eq. (4) uses $u^i_{\sigma_i} - l^i_{\sigma_i}$.

## §3.3 Basic Asymptotic Analysis

<a id="eq-5"></a>
### Eq. (5) — TBDA error decomposition
*§3.3, PDF p. 7*

$$
\frac{\tilde{\partial} f(\mathbf{x})}{\tilde{\partial} x_p} = \frac{\partial f(\mathbf{x})}{\partial x_{\sigma_i}} + O_P\!\left(\frac{1}{\sqrt{\eta(N)}}\right) + o(\operatorname{diam}(\mathbf{l}, \mathbf{u})). \tag{5}
$$

> ⚠️ **Transcription flags** (original kept as printed):
> - Left side uses $x_p$, right side $x_{\sigma_i}$; Theorem 1 (same page) uses $x_{\sigma_i}$ on both sides.

## §4.3 Convergence of Integro-Differential Quantities

<a id="eq-6"></a>
### Eq. (6) — Theorem 2: consistency of integrated TBDA quantities
*§4.3, PDF p. 8*

$$
\lim_{k \to \infty} \lim_{N \to \infty} \sum_{i \in \mathcal{D}_k} h(\tilde{\mathcal{D}}^i)\, \nu([\mathbf{l}_i, \mathbf{u}_i]) = \int_{[0,1]^P} h(\mathcal{D}f(\mathbf{x}))\, d\nu(\mathbf{x}) \,. \tag{6}
$$

> ⚠️ **Transcription flags** (original kept as printed):
> - Node set at depth $k$ is written $\mathcal{D}_k$ here but $\mathcal{I}_k$ in Algorithm 1; $\mathcal{D}$ is also the differential operator.
> - Node bounds written with subscripts $[\mathbf{l}_i, \mathbf{u}_i]$ here, superscripts $[\mathbf{l}^i, \mathbf{u}^i]$ in Eq. (4).

## §5.1 Introducing Tree-Based Models to PIML

<a id="eq-7"></a>
### Eq. (7) — PICART objective (tree analog of Eq. 3) ★
*§5.1, PDF p. 9*

$$
c(\hat{f}) = \sum_{i=1}^{N} (\hat{f}(\mathbf{x}_i) - y_i)^2 + \lambda_d \int_{\mathcal{X}} (\tilde{\mathcal{D}}\hat{f}(\mathbf{x}) - h(\hat{f}(\mathbf{x}), \mathbf{x}))^2 d\mathbf{x} + \lambda_b \int_{\mathcal{B}} (\hat{f}(\mathbf{x}) - b(\mathbf{x}))^2 d\mathbf{x} \,. \tag{7}
$$

[↑ index](#eq-index)

<a id="eq-8"></a>
### Eq. (8) — Classical CART split cost ★
*§5.1, PDF p. 9*

$$
\min_{\mu_l, \mu_r} \sum_{n \in \mathcal{N}_i} \left[ \mathbb{1}_{[x_{n,p} < \tau]} (y_n - \mu_l)^2 + \mathbb{1}_{[x_{n,p} \geq \tau]} (y_n - \mu_r)^2 \right] . \tag{8}
$$

> ⚠️ **Transcription flags** (original kept as printed):
> - Left child is $x_{n,p} < \tau$ here; §3.2 text defines the left side as $x_p \leq \tau_i$.

[↑ index](#eq-index)

<a id="eq-9"></a>
### Eq. (9) — Physics-penalized split cost ★
*§5.1, PDF p. 9*

$$
\sum_{n \in \mathcal{N}_i} \left[ \mathbb{1}_{[x_{n,p} < \tau]} (y_n - \mu_l)^2 + \mathbb{1}_{[x_{n,p} \geq \tau]} (y_n - \mu_r)^2 \right] + \lambda_p \left[ \operatorname{vol}([\mathbf{l}, \mathbf{u}]) \left( \tilde{\mathcal{D}}^i(\mu_l, \mu_r) - h\!\left( \mu, \frac{\mathbf{u}^i + \mathbf{l}^i}{2} \right) \right)^2 \right] . \tag{9}
$$

> ⚠️ **Transcription flags** (original kept as printed):
> - Penalty weight is $\lambda_p$ here, $\lambda_d$ in Eqs. (3)/(7); $p$ also indexes the split variable.
> - $\operatorname{vol}([\mathbf{l}, \mathbf{u}])$ has no node superscript $i$, unlike $\mathbf{u}^i, \mathbf{l}^i$ in the same term.
> - First argument of $h$ is an unsubscripted $\mu$; Algorithm 2 (p. 10) uses $\mu^{\rho_i}$.

## §5.2 First Order Differential Equations

<a id="eq-10"></a>
### Eq. (10) — First-order split cost (fixed $p$, $\tau$) ★
*§5.2, PDF p. 10*

$$
c(\mu_l^i, \mu_r^i, \tau, g) = N_l(\bar{y}_l - \mu_l^i)^2 + N_r(\bar{y}_r - \mu_r^i)^2 + \tilde{\lambda}\left(\frac{2(\mu_r^i - \mu_l^i)}{w} - \phi\right)^2 , \tag{10}

> - Weight printed as $\tilde{\lambda}$ (tilde); the text below and Eq. (11) use $\bar{\lambda}$ (bar).

### Eq. (11) — First-order closed-form child values ★
*§5.2, PDF p. 10*

$$
\begin{bmatrix} \mu_l^i \\ \mu_r^i \end{bmatrix} = \begin{bmatrix} N_r + \frac{4\bar{\lambda}}{w^2} & -\frac{4\bar{\lambda}}{w^2} \\ -\frac{4\bar{\lambda}}{w^2} & N_l + \frac{4\bar{\lambda}}{w^2} \end{bmatrix}^{-1} \begin{bmatrix} N_r \bar{y}_l - \frac{2\bar{\lambda}\phi}{w} \\ N_l \bar{y}_r + \frac{2\bar{\lambda}\phi}{w} \end{bmatrix} . \tag{11}

### Eq. (12) — Second-order ODE (1D)
*§5.3, PDF p. 11*

$$
by' + cy'' = f(x, y) \iff (by' + cy'' - f(x, y))^2 = 0 \,. \tag{12}

### Eq. (13) — Second-order auxiliary quantities $\alpha$, $z$ (left child)
*§5.3, PDF p. 11*

\alpha = \frac{2b}{u^i_p - l^i_p} - \frac{3c}{(u^i_p - \tau^i)(u^i_p - l^i_p)} \quad \text{and} \quad z = \frac{f\!\left(\frac{u^i_p + l^i_p}{2}, \mu^{\rho_i}\right) - \frac{3c\, g^{\rho_i}_p}{2(u^{\rho_i} - \tau^i)}}{\alpha} \,, \tag{13}

### Eq. (14) — Second-order closed-form child values
